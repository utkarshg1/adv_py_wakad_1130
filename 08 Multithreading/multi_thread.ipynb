# Multithreading

# Single threaded - One function can be executed at a time
# Multi threaded - Multiple functions can be executed at a time

```python
import time
```

```python
%%time
time.sleep(2)  # it will be 2 seconds idle
print("Hello")
time.sleep(3)  # it will be 3 secods idle
print("World!")
```

```python
def simple_intrest(p, n, r):
    print("Simple Intrest function started")
    time.sleep(2)
    i = (p * n * r) / 100
    a = p + i
    print(f"Simple Interest : {i:.2f} INR")
    print(f"Amount : {a:.2f} INR")
    print("Simple intrest function completed")
```

```python
%%time
simple_intrest(p=10000, n=5, r=6.5)
```

```python
import math


def hypotenuse(a, b):
    print("Hypotenuse function started")
    time.sleep(3)
    c = math.sqrt(a**2 + b**2)
    print(f"Hypotenuse of {a} and {b} is {c:.2f}")
    print("Hypotenuse function completed")
```

```python
%%time
hypotenuse(a=3, b=4)
```

# Single threading executes 1 function at a time

```python
%%time
simple_intrest(p=10000, n=5, r=4.5)  # 2 sec
hypotenuse(a=12, b=13)  # 3 sec
```

# Multithreading - execute multiple functions simultaneously

```python
%%time
from threading import Thread

# Create a thread for simple interst
th1 = Thread(target=simple_intrest, args=(20000, 5, 6.5))
th2 = Thread(target=hypotenuse, args=(3, 4))

# Start the thread
th1.start()
th2.start()

# Wait for all threads to finish
th1.join()
th2.join()
```

# Same function to calculate multiple values parallely

```python
def square(n):
    time.sleep(1)
    s = n**2
    print(f"Square of number {n} is {s}")
```

```python
%%time
square(11)
```

```python
a = [1, 3, 11, 15, 20, 21, 25, 27]
len(a)
```

```python
def square_of_list(a: list):
    for i in a:
        square(i)
```

```python
%%time
square_of_list(a)
```

```python
from threading import Thread


def square_multihtreading(a: list):
    # Create a blank list where i will start threads
    threads = []
    for i in a:
        th = Thread(target=square, args=(i,))
        th.start()
        threads.append(th)

    # Wait for all threads to finisth
    for th in threads:
        th.join()
```

```python
a
```

```python
%%time
square_multihtreading(a)
```

```python
b = list(range(1, 101))
print(b)
```

```python
%%time
square_multihtreading(b)
```

# Downloading files using single threading vs multihreading

```python
url1 = "https://raw.githubusercontent.com/utkarshg1/mlproject_regression/main/artifacts/data.csv"
print(url1)
```

```python
url1.split("/")
```

```python
url1.split("/")[-1]
```

```python
from urllib.request import urlretrieve


def download_file(url: str):
    filename = url.split("/")[-1]
    print(f"Downloading {filename} ...")
    urlretrieve(url, filename)
    print(f"{filename} download successfully")
```

```python
download_file(url1)
```

```python
urls = [
    "https://raw.githubusercontent.com/utkarshg1/mlproject_regression/main/artifacts/data.csv",
    "https://raw.githubusercontent.com/utkarshg1/mlproject_regression/main/artifacts/test.csv",
    "https://raw.githubusercontent.com/utkarshg1/mlproject_regression/main/artifacts/train.csv",
]
```

```python
def downloads_single_thread(urls: list[str]):
    for url in urls:
        download_file(url)
```

```python
%%time
downloads_single_thread(urls)
```

```python
def download_multithreaded(urls: list[str]):
    threads = []

    for url in urls:
        th = Thread(target=download_file, args=(url,))
        th.start()
        threads.append(th)

    for th in threads:
        th.join()
```

```python
%%time
download_multithreaded(urls)
```
